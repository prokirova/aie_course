# 02 Experiments: model comparison

Ноутбук воспроизводит сравнение baseline и нескольких ML-моделей. Основная команда обучения: `python -m src.train`.

```python
import json
from pathlib import Path
import pandas as pd

METRICS_PATH = Path('..') / 'artifacts' / 'metrics.json'
metrics = json.loads(METRICS_PATH.read_text(encoding='utf-8'))
pd.DataFrame(metrics['results']).sort_values('roc_auc', ascending=False)
```

```python
summary = {
    'best_model': metrics['best_model'],
    'transformer_model': metrics['transformer_model'],
    'transformer_device': metrics['transformer_device'],
    'rows': metrics['rows'],
    'embedding_sample_size': metrics['embedding_sample_size'],
    'positive_rate_full_dataset': metrics['positive_rate_full_dataset'],
    'positive_rate_embedding_sample': metrics['positive_rate_embedding_sample'],
}
summary
```

Финальная модель выбрана по ROC-AUC с дополнительным контролем F1. В текущем запуске лучше всего сработала LogisticRegression: она проще ансамблей, быстро отвечает в API и лучше ранжирует клиентов по риску.
